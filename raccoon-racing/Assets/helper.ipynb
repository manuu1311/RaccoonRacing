```python
from PIL import Image
import os
from pathlib import Path
```

```python
# Folder containing PNG images
folder_name = "Animations\sprites\DefineSprite_416_UIMainMenu"

# Output size
TARGET_SIZE = (500, 500)

folder = Path(folder_name)

for image_path in folder.glob("*.png"):
    with Image.open(image_path) as img:
        width, height = img.size

        # Coordinates for centered 500x500 crop
        top=605
        left=507
        right=left+TARGET_SIZE[0]
        bottom=top+TARGET_SIZE[1]

        # Crop ONLY the central 500x500 region
        cropped = img.crop((left, top, right, bottom))

        # Save (overwrite original)
        cropped.save(image_path)

        print(f"Cropped: {image_path}")
```
